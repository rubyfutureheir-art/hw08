### 聚餐費用試算器說明

#### 1. 文字轉數字與計算順序
在處理數據時，文字（字串）不能直接做數學運算。因此：
* 我們使用 `float()` 將帶有小數點的金額文字轉換成「浮點數」（小數），以便精確計算。
* 我們使用 `int()` 將人數轉換為「整數」。

**計算順序**：
1. 取得 **原始餐費**。
2. 算出 **服務費** = `原始餐費 * 服務費比例 (10%)`。
3. 計算 **應付總額** = `原始餐費 + 服務費 - 折價券金額`。
4. 計算 **每人分攤費用** = `應付總額 / 分攤人數`。
5. 比較 **每人費用** 是否大於 **個人預算**，得到一個布林值（`True` 或 `False`）。

#### 2. f-string 格式化輸出解釋
在程式中，我們會看到類似 `f"{variable:.2f}"` 的寫法，它的意思是：
* `f"..."`：這叫作 **格式化字串**（f-string），可以直接把變數放入大括號 `{}` 中顯示。
* `:.2f`：表示將數字格式化輸出為**小數點後兩位**（`f` 代表 float 浮點數，`2` 代表兩位小數）。例如 `1200.0` 會顯示成 `1200.00`，看起來更專業整齊！

```python
# ==========================================
# 1. 原始資料輸入 (字串型態)
# ==========================================
meal_cost_str = "1200.00"
service_charge_ratio = 0.10  # 10% 服務費
discount_coupon_str = "100.00"
people_count_str = "4"
budget_per_person_str = "300.00"

# ==========================================
# 2. 資料整理：文字轉換成數字
# ==========================================
# 將金額字串轉為浮點數 (float)
original_meal_cost = float(meal_cost_str)
discount_amount = float(discount_coupon_str)
budget_per_person = float(budget_per_person_str)

# 將人數轉為整數 (int)
people_count = int(people_count_str)

# ==========================================
# 3. 費用計算步驟
# ==========================================
# 服務費必須以原始餐費計算
service_charge = original_meal_cost * service_charge_ratio

# 應付總額 = 原始餐費 + 服務費 - 折價券
total_payable = original_meal_cost + service_charge - discount_amount

# 每人分攤費用 = 總額 / 人數
cost_per_person = total_payable / people_count

# 檢查是否超過預算 (布林值 True / False)
is_over_budget = cost_per_person > budget_per_person

# 計算超支金額
over_amount = cost_per_person - budget_per_person

# ==========================================
# 4. 格式化輸出結果
# ==========================================
print("=== 聚餐費用試算結果 ===")
print(f"聚餐名稱：weekend dinner")
print(f"原始餐費：{original_meal_cost:.2f} 元")
print(f"服務費：{service_charge:.2f} 元")
print(f"折價券：{discount_amount:.2f} 元")
print(f"應付總額：{total_payable:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")

if is_over_budget:
    print(f"（每人超支 {over_amount:.2f} 元）")
else:
    print("（在預算範圍內！）")
```

### 測試一：原始資料（人數為 4 人，折價券為 100 元）

```python
# 1. 原始資料：保留文字資料，另外建立整理後的變數。
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100.00
budget_per_person = 300.00

# 2. 整理聚餐名稱
clean_event_name = event_name.strip().lower()

# 3. 清理餐費文字，再轉換餐費與人數的型別
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").replace(" ", "")
meal_value = float(clean_meal_text)
people_count = int(people)

# 4. 先按原始餐費計算服務費，再扣除折價券
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count

# 5. 比較結果直接保存為 bool。等於或小於預算時不算超過
is_over_budget = amount_per_person > budget_per_person

# 6. 輸出時才將金額格式化為兩位小數
print("=== 測試一結果 ===")
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_value:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{amount_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
